```python
import numpy as np
learning_rate=0.01
ephocs=5
X=np.array([1,2,3,4,5])
Y=np.array([2,4,6,8,10])
m,b=0.0,1
n=len(X)

for i in range(ephocs):
  Y_pred=m*X+b
  cost=((Y-Y_pred)**2).mean()
  derivative_m=(-2/n)*sum(X*(Y-Y_pred))
  derivative_b=(-2/n)*sum(Y-Y_pred)
  m -=learning_rate * derivative_m
  b -=learning_rate * derivative_b
  print(f"Ephocs {i+1}/{ephocs} Cost {cost}  m {m} b {b}")
```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
X=np.array([[1],[3],[4],[5],[7]])
Y=np.array([3,5,7,9,11])
```

```python
plt.figure(figsize=(10,6))
# Correct X by extracting the numeric value from each set
X_corrected = np.array([list(item)[0] for item in X])
plt.scatter(X_corrected,Y,color='blue')
plt.title("Before appling linear regression")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
model=LinearRegression()
model.fit(X,Y)
y_pred=model.predict(X)
print(f"Predicted values {y_pred}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X,Y,color='blue',label="Origanal Data")
plt.plot(X,y_pred,color='red',label="Predicted Line")
plt.legend()
plt.title("After appling linear regression")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
m-model.coef_
b=model.intercept_

print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
```

```python
r2=model.score(X,Y)
print(f"R-squared (coefficient of determination): {r2}")

adjusted_r2 = 1 - (1 - r2) * ((len(Y) - 1) / (len(Y) - X.shape[1] - 1))
print(f"Adjusted R-squared: {adjusted_r2}")
#
```
