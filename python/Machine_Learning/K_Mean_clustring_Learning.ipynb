```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
X_data,_=make_blobs(
    n_samples=150,
    # n_features=2,
    centers=3,
    cluster_std=0.1,
    shuffle=True,
    random_state=42
)

k_values=range(2,15)

wcss=[]  # Within cluster
sill_score=[]


for k in k_values:
    kmeans=KMeans(n_clusters=k,random_state=42)
    labels=kmeans.fit_predict(X_data)
    wcss.append(kmeans.inertia_)
    sill_score.append(silhouette_score(X_data,labels))

wcss

```

```python
plt.figure(figsize=(10,6))
plt.plot(k_values,wcss,marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method for Optimal K")
plt.grid(True)
plt.show()



plt.figure(figsize=(10,6))
plt.plot(k_values,sill_score,marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Optimal K")
plt.grid(True)
plt.show()
```

```python
best_k=np.argmax(sill_score)
print(f"Best K according to silhouette Score {best_k}")
best_k_w=np.argmin(wcss)
print(f"Best K according to WCSS {best_k_w}")
```

Second Lecture

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


X,_=make_blobs(
    n_samples=200,
    centers=3,
    cluster_std=1.0,
    shuffle=True,
    random_state=42
)

kmeans=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=kmeans.fit_predict(X)

plt.figure(figsize=(6,5))
plt.scatter(X[:,0],X[:,1],c=labels,cmap="tab10",alpha=0.7)
plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    marker="X",
    s=200,
    color="black"
)
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.show()

```

```python
print(f"Intera(Sum of Squered Distance)")
print(round(kmeans.inertia_,2))
print(f"Silhouette Score")
print(round(silhouette_score(X,labels),2))
```

```python
for k_test in [1,2,3,4,5,6]:
  km=KMeans(n_clusters=k_test,random_state=42,n_init=10)
  labels=km.fit_predict(X)
  print(f"K={k_test} inertia : {round(km.inertia_,2)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X,_=make_blobs(
    n_samples=200,
    centers=3,
    cluster_std=1.0,
    shuffle=True,
    random_state=42

)
kmeans=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=kmeans.fit_predict(X)

plt.figure(figsize=(6,5))
plt.scatter(X[:,0],X[:,1],c=labels,cmap='tab10')
plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    marker='X',
    s=200,
    color='black'
)
plt.title('K-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()

```

```python
print("Intertia (Sum of Squared Distances):")
print(round(kmeans.inertia_, 2))
print("Silhouette Score:")
print(round(silhouette_score(X, labels), 2))

for best_k in [1,2,3,4]:
  km=KMeans(n_clusters=best_k,random_state=42,n_init=10)
  labels=km.fit_predict(X)
  print(f"K={best_k} inertia : {round(km.inertia_,2)}")
```
