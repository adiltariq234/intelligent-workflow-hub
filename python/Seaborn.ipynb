```python
numric_df=df[
    ['length_of_stay_days','treatment_cost','recovery_score','readmission_risk']
]
corr_matrix=numric_df.corr()
corr_matrix
```

```python
sns.heatmap(
    data=corr_matrix,
    annot=True,
    cmap='coolwarm',

)
plt.title("Correlation Heatmap")
plt.show()
#
```

```python
import numpy as np

mask=np.triu(np.ones_like(corr_matrix,dtype=bool))
sns.heatmap(
    data=corr_matrix,
    annot=True,
    cmap='coolwarm',
    mask=mask
)
plt.title("Correlation Heatmap")
plt.show()
```
