```python
class student:
  def __init__(self, name, grade):
    self.name = name
    self._grade = grade # Use _grade for internal storage of the property

  @property
  def grade(self):
    return self._grade

  @grade.setter
  def grade(self, new_grade):
    if 0 <= new_grade <= 100:
      self._grade = new_grade
    else:
      print("Invalid grade")

student1 = student("Adil", 90)
print(student1.grade)
student1.grade = 95
print(student1.grade)
student1.grade = 105 # This will print 'Invalid grade'
```

```python
import math


class Point:
    def __init__(self, x, y):
        # Har object ke apne x aur y coordinates honge.
        self.x = x
        self.y = y

    def __str__(self):
        # print(point) karne par yeh text dikhai dega.
        return f"Point({self.x}, {self.y})"

    def __add__(self, other):
        # Sirf Point + Point allow karna hai.
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)

    def __len__(self):
        # Is project me har Point ke 2 coordinates hain: x aur y.
        return 2

    def distance_from_origin(self):
        # Origin (0, 0) se distance nikalne ka formula.
        return math.sqrt(self.x ** 2 + self.y ** 2)

    def distance_from(self, other):
        if not isinstance(other, Point):
            raise TypeError("distance_from() sirf Point object ke sath chalega.")
        return math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2)


def make_point(point_name):
    print(f"\n{point_name} ke coordinates enter karein:")
    x = float(input("x: "))
    y = float(input("y: "))
    return Point(x, y)


def show_result(p1, p2):
    print("\n--- Result ---")
    print("Pehla point:", p1)
    print("Doosra point:", p2)

    added_point = p1 + p2
    print("Addition (p1 + p2):", added_point)

    print("len(p1):", len(p1))
    print("P1 ka distance origin se:", round(p1.distance_from_origin(), 2))
    print("P2 ka distance origin se:", round(p2.distance_from_origin(), 2))
    print("P1 aur P2 ke darmiyan distance:", round(p1.distance_from(p2), 2))


def main():
    print("Point Class Project")
    print("Is program me hum Point class ka use karke OOP aur magic methods samjhenge.")

    p1 = make_point("Point 1")
    p2 = make_point("Point 2")

    show_result(p1, p2)


if __name__ == "__main__":
    main()
```
