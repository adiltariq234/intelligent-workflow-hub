```python
duplicated_rows=df.columns.difference(['student_id'])
num_duplicated_rows=df.duplicated(subset=duplicated_rows).sum()

print(f"\nNumber Of Duplicated row {num_duplicated_rows}")

if num_duplicated_rows >0:
  print("\n Duplicated Row")
  print(df[df.duplicated(subset=duplicated_rows,keep=False)])
```

```python
duplicated_rows=df.columns.difference(['student_id'])
rows_before=len(df)
df.drop_duplicates(subset=duplicated_rows,keep="first",inplace=True)
rows_removed=rows_before - len(df)
print(f"\n{rows_removed} duplicated rows removed")

print(f"\nNumber Of Duplicated row {num_duplicated_rows}")

```

```python
df
```

```python
df.dropna(subset=['name'],inplace=True)
```

```python
df['attendance_percentage'] = pd.to_numeric(df['attendance_percentage'], errors='coerce')
attendance_mean = df['attendance_percentage'].mean()
df['attendance_percentage'] = df['attendance_percentage'].fillna(attendance_mean)
```

```python
df
```

```python
print("\nMissing value after Cleaning")
print("="*100)
missing_value=df.isnull().sum()
print(missing_value)
print(f"Missing value {missing_value.sum()}")
```

```python
print("Before Fixing Grade Cloumn")
print(df[['name','grade']].head(15))

df['grade']=pd.to_numeric(df['grade'],errors='coerce')
grade_median=df['grade'].median()
df['grade']=df['grade'].fillna(grade_median)

print("After Fixing Grade Cloumn")
print(df[['name','grade']].head())
```

```python
print("Students with impossible ages")
print("="*100)

# Step 1: Find impossible ages
impossible_ages = df[(df['age'] < 0) | (df['age'] > 100)]
print(f"Impossible Ages Count: {len(impossible_ages)}")

# Step 2: Get valid ages (18 to 100)
valid_ages = df[(df['age'] >= 18) & (df['age'] <= 100)]['age']

# Step 3: Median calculate
median_age = valid_ages.median()

# Step 4: Replace invalid ages with median
df.loc[(df['age'] < 0) | (df['age'] > 100), 'age'] = median_age

print("\nAfter Fixing Age Column")
print(df[['name', 'age']].head())
```

```python
print("Before Cleaning Gender value")
print(df['gender'].unique())
```

```python
#Standradization

df['gender']=df['gender'].str.strip().str.lower()

 # Corrected typo from 'stripe' to 'strip'

gender_maping={
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female",
    "transgender": "Transgender"
}

df['gender']=df['gender'].replace(gender_maping)
print("After Cleaning Gender value")
print()
print(df['gender'].unique())
```
