### Example 2: Iterating through a list of product details

```python
products_list = [
  {"ProductID": "P001", "Name": "Laptop", "Price": 1200, "Category": "Electronics"},
  {"ProductID": "P002", "Name": "Mouse", "Price": 25, "Category": "Electronics"},
  {"ProductID": "P003", "Name": "Keyboard", "Price": 75, "Category": "Electronics"}
]

print("--- Product Inventory ---")
for product in products_list:
  print(f"ID: {product.get('ProductID','N/A')} | Item: {product.get('Name','N/A')} | Price: ${product.get('Price','N/A')} | Category: {product.get('Category','N/A')}")
print("-------------------------")
```

```python
print("--- Products Priced Above $100 ---")
for product in products_list:
  if product.get('Price', 0) > 100:
    print(f"ID: {product.get('ProductID','N/A')} | Item: {product.get('Name','N/A')} | Price: ${product.get('Price','N/A')} | Category: {product.get('Category','N/A')}")
print("----------------------------------")
```
